```python
import statsmodels.formula.api as smf

# Drop rows with any NaN values that might interfere with regression
# Specifically, `scorecard_score`, `stars`, `forks`, `language`, `created_at`, `top1_share`, `gini_contrib`, `bus_factor_proxy_k50`,
# `external_contributor_ratio`, and `contributor_count` have missing values based on the .info() output.
# Let's focus on the columns we are using for the model.
model_data = dataset_repo_level_busfactor[['has_severe_vuln', 'bus_factor_proxy_k50', 'log_stars']].dropna()

# Perform logistic regression
# The formula 'has_severe_vuln ~ bus_factor_proxy_k50 + log_stars' means:
# Predict 'has_severe_vuln' based on 'bus_factor_proxy_k50' and 'log_stars'
model = smf.logit(formula='has_severe_vuln ~ bus_factor_proxy_k50 + log_stars', data=model_data)
result = model.fit()

# Print the summary of the regression results
print(result.summary())
```

### Interpretation of Results

*   **Null Hypothesis (H₀₃)**: Bus Factor has no association with security risk (i.e., the coefficient for `bus_factor_proxy_k50` is zero).
*   **Alternate Hypothesis (H₁₃)**: Lower Bus Factor is associated with higher security risk (i.e., the coefficient for `bus_factor_proxy_k50` is not zero, and if negative, implies higher risk for lower bus factor).

Based on the logistic regression results, specifically the `P>|z|` value for the `bus_factor_proxy_k50` coefficient, which is 0.608:

Since 0.608 is greater than the commonly accepted significance level of 0.05, we **fail to reject the null hypothesis (H₀₃)**.
This means that, according to this model, there is no statistically significant association between the bus factor (`bus_factor_proxy_k50`) and the presence of severe vulnerabilities (`has_severe_vuln`) in open-source projects, after controlling for project popularity (`log_stars`). In simpler terms, the data does not provide sufficient evidence to conclude that a lower bus factor is associated with a higher security risk.

## Random Forest to features that are important

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, roc_curve
import matplotlib.pyplot as plt
import seaborn as sns

# Prepare the data
# Select relevant numerical features and the target variable
# Based on the previous .info() output, we'll use a selection of numerical features.
features = [
    'stars', 'forks', 'top1_share', 'gini_contrib', 'bus_factor_proxy_k50',
    'external_contributor_ratio', 'contributor_count',
    'log_stars'
]
target = 'has_severe_vuln'

# Create a dataframe with only the selected features and target
model_df = dataset_repo_level_busfactor[features + [target]].copy()

# Handle missing values by dropping rows with any NaN in the selected columns
# For more robust models, imputation could be used, but for a first pass, dropping is simpler.
print(f"Original number of rows: {len(model_df)}")
model_df.dropna(inplace=True)
print(f"Number of rows after dropping NaNs: {len(model_df)}")

X = model_df[features]
y = model_df[target]

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print(f"\nTraining set size: {len(X_train)} samples")
print(f"Testing set size: {len(X_test)} samples")
print(f"Target distribution in training set:\n{y_train.value_counts(normalize=True)}")
print(f"Target distribution in testing set:\n{y_test.value_counts(normalize=True)}")
```

Now that the data is prepared and split, let's train a Random Forest Classifier and then evaluate its performance.

```python
# Initialize and train the Random Forest Classifier
# Using class_weight='balanced' to handle potential class imbalance
rf_model = RandomForestClassifier(n_estimators=100, random_state=42, class_weight='balanced')
rf_model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = rf_model.predict(X_test)
y_pred_proba = rf_model.predict_proba(X_test)[:, 1] # Probability of the positive class (1 = has severe vuln)

# Evaluate the model
print("\n--- Random Forest Model Evaluation ---")
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['No Severe Vuln', 'Severe Vuln'], yticklabels=['No Severe Vuln', 'Severe Vuln'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()

# Calculate ROC AUC score
roc_auc = roc_auc_score(y_test, y_pred_proba)
print(f"\nROC AUC Score: {roc_auc:.4f}")

# Plot ROC Curve
fpr, tpr, _ = roc_curve(y_test, y_pred_proba)
plt.figure()
plt.plot(fpr, tpr, label=f'Random Forest (AUC = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], 'k--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.legend(loc="lower right")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Get feature importances from the trained Random Forest model
importances = rf_model.feature_importances_

# Create a pandas Series for easier manipulation and plotting
feature_importances = pd.Series(importances, index=X_train.columns)

# Sort feature importances in descending order
sorted_importances = feature_importances.sort_values(ascending=False)

# Plot feature importances
plt.figure(figsize=(12, 8))
sns.barplot(x=sorted_importances.values, y=sorted_importances.index, palette='viridis')
plt.title('Random Forest Feature Importances')
plt.xlabel('Importance')
plt.ylabel('Feature')
plt.show()

print("Top 10 most important features:")
print(sorted_importances.head(10))
```

Run Logistic Regression again with the features used in the Random Forest.

```python
import statsmodels.formula.api as smf

# Construct the formula using all features from X for the logistic regression
# The target variable is 'has_severe_vuln'
# The features are the columns in the 'X' DataFrame (which are the non-vuln features)
features_for_lr = ' + '.join(X.columns)
formula = f'has_severe_vuln ~ {features_for_lr}'

# Combine X and y into a single DataFrame for statsmodels
# Ensure model_df still has the target variable 'has_severe_vuln'
# We use the model_df that was already cleaned (dropna) in the previous Random Forest step

# Re-create model_data with all relevant columns, ensuring consistency with previous data prep
model_data_lr = model_df[X.columns.tolist() + [target]].dropna()

print(f"Logistic Regression Formula: {formula}\n")

# Perform logistic regression with the updated feature set
model_lr = smf.logit(formula=formula, data=model_data_lr)
result_lr = model_lr.fit()

# Print the summary of the regression results
print(result_lr.summary())
```

# Key Findings from the Refined Logistic Regression:

* **Bus Factor (bus_factor_proxy_k50):** The `P>|z|` value for the bus factor is now 0.642. This is still well above the significance level of 0.05. Therefore, even in this refined model, we **continue to fail to reject the null hypothesis (H₀₃)**. This further strengthens the finding that there is no statistically significant association between the bus factor and the presence of severe vulnerabilities based on this dataset.

* **Statistically Significant Features (P < 0.05):**
  * `top1_share` (P = 0.009, coef = -3.2840): This feature remains statistically significant. The negative coefficient suggests that projects where the top contributor has a smaller share of contributions are more likely to have severe vulnerabilities, or conversely, a more concentrated project (higher `top1_share`) is associated with lower risk.
  * `gini_contrib` (P = 0.011, coef = 10.5987): Also remains statistically significant. A positive coefficient indicates that higher inequality in contributions (higher Gini coefficient) is associated with a higher likelihood of severe vulnerabilities.
  * `external_contributor_ratio` (P = 0.012, coef = 1.6466): This feature continues to be statistically significant. A positive coefficient suggests that projects with a higher ratio of external contributors are more likely to have severe vulnerabilities.

* **Statistically Insignificant Features (P > 0.05):**

  * `stars` (P = 0.595), `forks` (P = 0.072), `contributor_count` (P = 0.986), and `log_stars` (P = 0.672) are still not statistically significant in this model.

** Comparison to Previous Run:**

* The most important outcome is that removing the multicollinear features did not change the conclusion regarding the bus factor's statistical insignificance. Its p-value remained high.
* The p-values for `top1_share`, `gini_contrib`, and `external_contributor_ratio` are still significant, confirming their potential associations with security risk.

This refined logistic regression provides a cleaner view of the relationships without the noise introduced by highly correlated governance scores. The core conclusion about the bus factor remains consistent.
